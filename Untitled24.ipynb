```python
plt.figure(figsize=(8,8))
sns.boxplot(x=df['snr'])
plt.title("Detection of outliers in snr")
plt.show()
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(x=df['peak_frequency'] , color = 'skyblue')
plt.title("Detection of outliers in peak_frequency")
plt.show()
```

```python
#The Boxplots for snr and peak_frequency initially showed a significant number of extreme values (outliers) that could skew the model's performance, I applied the Interquartile Range (IQR) method to filter these points.
```

```python
columns = ['snr', 'peak_frequency']
for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3-Q1
    lower_bound = Q1-1.5*IQR
    upper_bound = Q3+1.5*IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(x=df['snr'])
plt.title("Detection of outliers in snr")
plt.show()
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(x=df['peak_frequency'] , color = 'skyblue')
plt.title("Detection of outliers in peak_frequency")
plt.show()
```

```python
#This process refined our dataset from 7,966 to 5,677 rows, ensuring that the model learns from the most representative signal patterns of gravity waves and glitches
```
